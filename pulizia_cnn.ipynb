```python
# Carico i dati
import pandas as pd

df_cnn = pd.read_csv("CNN_Articels.csv", sep=",", encoding="utf-8")
```

```python
# Informazioni generali sul dataframe
print(df_cnn.dtypes)
print(df_cnn.shape[0])
```

```python
# calcolo i valori mancanti per ogni colonna
missing_count_cnn = {}
for col in df_cnn.columns:
    missing_count = df_cnn[col].isnull().sum()
    missing_count_cnn[col] = missing_count
    print(f"Colonna: {col}, Valori mancanti: {missing_count}")
```

```python
# Analizzo le righe duplicate per Url e Article text insieme
duplicate_count_cnn = df_cnn.duplicated(subset=["Url", "Article text"]).sum()
print(f"Righe duplicate (stesso Url e stesso Article text): {duplicate_count_cnn}")

# rimuovo le righe duplicate solo se Url e Article text sono uguali, mantenendo la prima occorrenza
df_cnn = df_cnn.drop_duplicates(subset=["Url", "Article text"], keep="first")
```

```python
# Tengo solo le colonne utili per l'analisi
df_cnn = df_cnn[["Author", "Date published", "Category", "Section", "Headline", "Article text"]]

# Salvo il dataframe pulito in un nuovo file CSV
df_cnn.to_csv("CNN_articles_cleaned.csv", index=False, encoding="utf-8")
```

```python
# vedo i valori unici della colonna Section
sezione_cnn = df_cnn['Section'].unique()
category_cnn = df_cnn['Category'].unique()

print(sezione_cnn)
print(category_cnn)
```

```python
# raccolgo tutte le variabili calcolate durante l'analisi in un unico dizionario,
# cosi' da poterle riutilizzare facilmente in un altro notebook
extra_cnn = {
    "missing_count": missing_count_cnn,
    "duplicate_count": duplicate_count_cnn,
    "sezioni": sezione_cnn,
    "categorie": category_cnn,
}

#salvo il dizionario in un file json all'interno della cartella del progetto, cosi' da poterlo riutilizzare in un altro notebook
```

```python
import json
import numpy as np


def _json_default(obj):
    """Converte in tipi standard i valori non nativamente serializzabili in JSON
    (Timestamp di pandas, DataFrame, array/interi/float numpy)."""
    if isinstance(obj, pd.DataFrame):
        return obj.to_dict(orient="records")
    if isinstance(obj, pd.Series):
        return obj.to_dict()
    if isinstance(obj, pd.Timestamp):
        return obj.isoformat()
    if isinstance(obj, np.ndarray):
        return obj.tolist()
    if isinstance(obj, np.integer):
        return int(obj)
    if isinstance(obj, np.floating):
        return float(obj)
    raise TypeError(f"Tipo non serializzabile in JSON: {type(obj)}")


with open("extra_cnn.json", "w", encoding="utf-8") as f:
    json.dump(extra_cnn, f, ensure_ascii=False, indent=2, default=_json_default)

print("Salvato extra_cnn.json")
```
